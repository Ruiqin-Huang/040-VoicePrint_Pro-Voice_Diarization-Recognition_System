```python
def show_clusters(clusters):
    # 提取坐标和类别
    x = np.array([item['x_coordinate'] for item in clusters])
    y = np.array([item['y_coordinate'] for item in clusters])
    labels = np.array([item['cluster_id'] for item in clusters])
    files = [item['segment_audio_file'] for item in clusters]

    # 自动生成颜色，支持1~100类
    num_classes = len(set(labels))
    cmap = plt.get_cmap('tab20' if num_classes <= 20 else 'tab20b')
    colors = [cmap(l % cmap.N) for l in labels]

    plt.figure(figsize=(10, 8))
    scatter = plt.scatter(x, y, c=labels, cmap=cmap, s=80, edgecolors='k', alpha=0.7)

    # 标注文件名
    for xi, yi, fname in zip(x, y, files):
        plt.text(xi, yi, fname, fontsize=8, ha='right', va='bottom', alpha=0.7)

    plt.title(f'Cluster Visualization ({num_classes} classes)')
    plt.xlabel('t-SNE X')
    plt.ylabel('t-SNE Y')
    plt.grid(True, linestyle='--', alpha=0.3)
    plt.colorbar(scatter, label='Cluster ID')
    plt.tight_layout()
    plt.show()

# 调用可视化
show_clusters(CLUSTERS_ERC)
```
